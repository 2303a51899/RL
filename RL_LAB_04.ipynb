```python
import math

machines = ['A', 'B', 'C']

M = 3
N_play = 10

N = {machine: 0 for machine in machines}
R = {machine: 0 for machine in machines}
Q = {machine: 0.0 for machine in machines}
UCB = {machine: float('inf') for machine in machines}

total_reward = 0

print("Upper Confidence Bound Multi-Armed Bandit")
print("------------------------------------------")

for play in range(1, N_play + 1):

    for machine in machines:
        if N[machine] == 0:
            UCB[machine] = float('inf')
        else:
            UCB[machine] = Q[machine] + math.sqrt(
                (2 * math.log(play)) / N[machine]
            )

    max_ucb = max(UCB.values())

    best_machines = [
        machine for machine in machines
        if UCB[machine] == max_ucb
    ]

    selected_machine = best_machines[0]

    print(f"\nPlay {play}")
    print(f"UCB values: A = {UCB['A']:.2f}, "
          f"B = {UCB['B']:.2f}, "
          f"C = {UCB['C']:.2f}")

    print(f"Play with machine {selected_machine}")

    reward = float(input("Enter reward: "))

    N[selected_machine] += 1
    R[selected_machine] += reward

    Q[selected_machine] = (
        Q[selected_machine]
        + (1 / N[selected_machine])
        * (reward - Q[selected_machine])
    )

    total_reward += reward

    print(f"Updated Q({selected_machine}) = "
          f"{Q[selected_machine]:.2f}")


print("\n\n========== FINAL RESULTS ==========")

print("\nMachine\tN\tR\tUCB\tQ")
print("---------------------------------------")

for machine in machines:
    print(
        f"{machine}\t"
        f"{N[machine]}\t"
        f"{R[machine]:.2f}\t"
        f"{UCB[machine]:.2f}\t"
        f"{Q[machine]:.2f}"
    )

print("\nTotal Reward =", total_reward)

print("\nAnalysis:")

for machine in machines:
    print(
        f"Machine {machine}: selected {N[machine]} times, "
        f"total reward = {R[machine]:.2f}, "
        f"UCB = {UCB[machine]:.2f}, "
        f"Q-value = {Q[machine]:.2f}"
    )
```
