```python
import random

machines = ['A', 'B', 'C']
epsilon = 0.2
plays = 10

N = {'A': 0, 'B': 0, 'C': 0}
R = {'A': [], 'B': [], 'C': []}
Q = {'A': 0.0, 'B': 0.0, 'C': 0.0}

total_reward = 0

for play in range(1, plays + 1):
    r = random.random()

    if r < epsilon:
        selected = random.choice(machines)
        action = "Exploration"
    else:
        selected = max(Q, key=Q.get)
        action = "Exploitation"

    print(f"\nPlay {play}")
    print(f"Random number: {r:.2f}")
    print(f"{action}")
    print(f"Play with machine {selected}")

    reward = float(input("Enter reward: "))

    N[selected] += 1
    R[selected].append(reward)
    Q[selected] = Q[selected] + (reward - Q[selected]) / N[selected]

    total_reward += reward

    print(f"N({selected}) = {N[selected]}")
    print(f"Q({selected}) = {Q[selected]:.2f}")

print("\n--- Final Results ---")

for machine in machines:
    print(f"\nMachine {machine}")
    print(f"N({machine}) = {N[machine]}")
    print(f"Rewards = {R[machine]}")
    print(f"Q({machine}) = {Q[machine]:.2f}")

print(f"\nTotal Reward after 10 plays = {total_reward}")
```
