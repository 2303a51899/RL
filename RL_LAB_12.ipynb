```python
gamma = 0.9
theta = 0.9

states = ['S0', 'S1', 'G']
non_terminal_states = ['S0', 'S1']
actions = ['Up', 'Down', 'Left', 'Right']

step_reward = -1.0
wall_reward = -1.0
goal_reward = 100.0

transitions = {
    'S0': {
        'Up': ('S0', wall_reward),
        'Down': ('S1', step_reward),
        'Left': ('S0', wall_reward),
        'Right': ('S0', wall_reward),
    },
    'S1': {
        'Up': ('S0', step_reward),
        'Down': ('S1', wall_reward),
        'Left': ('S1', wall_reward),
        'Right': ('G', goal_reward),
    },
}

print('=== 1. MDP DEFINITION ===')
print(f'States              : {states}')
print(f'Non-Terminal States : {non_terminal_states}')
print(f'Actions             : {actions}')
print(
    f'Rewards             : Step={step_reward}, Wall={wall_reward},'
    f' Goal={goal_reward}'
)
print(f'Discount Factor     : {gamma}\n')

print('=== TRANSITION TABLE ===')
for s in non_terminal_states:
  for a in actions:
    next_s, r = transitions[s][a]
    print(f'State: {s:<3} | Action: {a:<6} -> Next State: {next_s:<3} | Reward: {r}')
print()

V = {s: 0.0 for s in states}
print('=== 2. INITIAL STATE VALUES ===')
print(V, '\n')

print('=== 3, 4 & 5. VALUE ITERATION ===')
iteration = 0
while True:
  iteration += 1
  delta = 0.0
  V_new = V.copy()

  for s in non_terminal_states:
    q_values = [
        r + gamma * V[next_s] for next_s, r in transitions[s].values()
    ]
    V_new[s] = max(q_values)
    delta = max(delta, abs(V_new[s] - V[s]))

  V = V_new
  print(
      f'Iter {iteration:02d} | V(S0) = {V["S0"]:>8.4f} | V(S1) = {V["S1"]:>8.4f} |'
      f' Delta = {delta:.6f}'
  )

  if delta < theta:
    break
print()

Q = {s: {} for s in non_terminal_states}
optimal_policy = {}

for s in non_terminal_states:
  for a in actions:
    next_s, r = transitions[s][a]
    Q[s][a] = r + gamma * V[next_s]
  optimal_policy[s] = max(Q[s], key=Q[s].get)

print('=== FINAL OPTIMAL STATE VALUES V*(s) ===')
for s in states:
  print(f'V*({s}) = {V[s]:.4f}')
print()

print('=== FINAL ACTION-VALUE TABLE Q*(s, a) ===')
for s in non_terminal_states:
  for a in actions:
    next_s, _ = transitions[s][a]
    print(f'State: {s:<3} | Action: {a:<6} | Next: {next_s:<3} | Q*(s, a) = {Q[s][a]:.4f}')
print()

print('=== 6. OPTIMAL POLICY ===')
for s in non_terminal_states:
  print(f'State: {s:<3} | Optimal Value: {V[s]:>7.2f} | Optimal Action: {optimal_policy[s]}')
print(f'State: G   | Optimal Value: {V["G"]:>7.2f} | Optimal Action: Terminal')
```
