```python
import numpy as np

GAMMA = 0.9
POLICY = {
    "S0": "Right",
    "S1": "Down",
    "S3": "Right",
    "S4": "Up",
}


def step(state, action):
    transitions = {
        "S0": {"Up": "S0", "Down": "X", "Left": "S0", "Right": "S1"},
        "S1": {"Up": "S1", "Down": "S3", "Left": "S0", "Right": "G"},
        "S3": {"Up": "S1", "Down": "S3", "Left": "X", "Right": "S4"},
        "S4": {"Up": "G", "Down": "S4", "Left": "S3", "Right": "S4"},
    }

    next_state = transitions[state][action]

    if next_state == "G":
        return next_state, 10.0, True
    elif next_state == "X":
        return next_state, -5.0, True
    else:
        return next_state, -1.0, False


def state_value_function(start_state="S0"):
    curr_state = start_state
    discounted_return = 0.0
    discount = 1.0

    while True:
        action = POLICY[curr_state]
        next_state, reward, done = step(curr_state, action)
        discounted_return += discount * reward
        discount *= GAMMA
        if done:
            break
        curr_state = next_state

    return discounted_return


def action_value_function(start_state="S0", first_action="Up"):
    curr_state, reward, done = step(start_state, first_action)
    discounted_return = reward
    discount = GAMMA

    while not done:
        action = POLICY[curr_state]
        next_state, reward, done = step(curr_state, action)
        discounted_return += discount * reward
        discount *= GAMMA
        curr_state = next_state

    return discounted_return


v_s0 = state_value_function("S0")
q_s0_up = action_value_function("S0", "Up")

print(f"State Value V(S0): {v_s0:.4f}")
print(f"Action Value Q(S0, 'Up'): {q_s0_up:.4f}")
```
