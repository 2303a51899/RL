```python
class GridWorldEnv:
    def __init__(self):
        # Standard small deterministic gridworld transition dynamics
        # S0 -> Right -> S1 (reward 0)
        # S0 -> Up -> S0 (wall, reward -1)
        # S1 -> Down -> S4 (reward 0)
        # S3 -> Right -> G (Goal, reward +10)
        # S4 -> Up -> S3 (reward 0)
        self.transitions = {
            ('S0', 'Right'): ('S1', 0.0, False),
            ('S0', 'Up'): ('S0', -1.0, False),
            ('S1', 'Down'): ('S4', 0.0, False),
            ('S3', 'Right'): ('G', 10.0, True),
            ('S4', 'Up'): ('S3', 0.0, False)
        }

    def step(self, state, action):
        if (state, action) in self.transitions:
            return self.transitions[(state, action)]
        # Default fallback if transition not explicitly defined
        return state, -1.0, False

env = GridWorldEnv()

policy = {
    'S0': 'Right',
    'S1': 'Down',
    'S3': 'Right',
    'S4': 'Up',
}

gamma = 0.9


def state_value_function(env, policy, start_state, gamma=0.9, max_steps=100):
  state = start_state
  total_return = 0.0
  discount = 1.0

  for _ in range(max_steps):
    if state in ['G', 'X']:
      break
    action = policy[state]
    next_state, reward, done = env.step(state, action)
    total_return += discount * reward
    discount *= gamma
    state = next_state
    if done:
      break

  return total_return


def action_value_function(
    env, policy, start_state, first_action, gamma=0.9, max_steps=100
):
  if start_state in ['G', 'X']:
    return 0.0

  next_state, reward, done = env.step(start_state, first_action)
  total_return = reward

  if not done and next_state not in ['G', 'X']:
    total_return += gamma * state_value_function(
        env, policy, next_state, gamma, max_steps - 1
    )

  return total_return


v_s0 = state_value_function(env, policy, 'S0', gamma)
q_s0_up = action_value_function(env, policy, 'S0', 'Up', gamma)

print(f'State Value V(S0)         : {v_s0:.4f}')
print(f'Action Value Q(S0, Up)    : {q_s0_up:.4f}')
```
