```python
import numpy as np

_grid_transitions = {
    ("S0", "Right"): ("S1", -1),
    ("S0", "Down"): ("S3", -1),
    ("S0", "Up"): ("X", -10),
    ("S0", "Left"): ("X", -10),

    ("S1", "Right"): ("G", 10),
    ("S1", "Down"): ("S4", -1),
    ("S1", "Up"): ("X", -10),
    ("S1", "Left"): ("S0", -1),

    ("S3", "Right"): ("S4", -1),
    ("S3", "Down"): ("X", -10),
    ("S3", "Up"): ("S0", -1),
    ("S3", "Left"): ("X", -10),

    ("S4", "Right"): ("X", -10),
    ("S4", "Down"): ("X", -10),
    ("S4", "Up"): ("S1", -1),
    ("S4", "Left"): ("S3", -1)
}

_terminal_states = ["G", "X"]

def step(state, action):
    if state in _terminal_states:
        return state, 0

    return _grid_transitions.get((state, action), (state, -100))


policy = {
    "S0": "Right",
    "S1": "Down",
    "S3": "Right",
    "S4": "Up"
}

gamma = 0.9

def state_value_function(state):
    total = 0
    discount = 1
    current_state = state

    max_steps = 100
    steps_taken = 0

    while current_state not in ["G", "X"] and steps_taken < max_steps:
        action = policy[current_state]
        next_state, reward = step(current_state, action)

        total += discount * reward
        discount *= gamma
        current_state = next_state
        steps_taken += 1

    return total


def action_value_function(state, action):
    next_state, reward = step(state, action)

    if next_state == "G" or next_state == "X":
        return reward
    else:
        return reward + gamma * state_value_function(next_state)


print("State Value of S0:", state_value_function("S0"))
print("Action Value of S0 and Up:", action_value_function("S0", "Up"))
```
