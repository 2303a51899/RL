```python
import random

machines = ['A', 'B', 'C']

M = 3
N_play = 10
epsilon = 0.2

N = {machine: 0 for machine in machines}
R = {machine: 0 for machine in machines}
Q = {machine: 0.0 for machine in machines}

total_reward = 0

print("Modified Epsilon-Greedy Multi-Armed Bandit")
print("-------------------------------------------")

for play in range(1, N_play + 1):

    r = random.random()

    if r < epsilon:
        max_q = max(Q.values())

        non_best_machines = [
            machine for machine in machines
            if Q[machine] < max_q
        ]

        if not non_best_machines:
            selected_machine = random.choice(machines)
        else:
            selected_machine = random.choice(non_best_machines)

        decision = "Exploration"

    else:
        max_q = max(Q.values())

        best_machines = [
            machine for machine in machines
            if Q[machine] == max_q
        ]

        selected_machine = random.choice(best_machines)
        decision = "Exploitation"

    print(f"\nPlay {play}")
    print(f"Random number = {r:.4f}")
    print(f"Decision = {decision}")
    print(f"Play with machine {selected_machine}")

    reward = float(input("Enter reward: "))

    N[selected_machine] += 1
    R[selected_machine] += reward

    Q[selected_machine] = (
        Q[selected_machine]
        + (1 / N[selected_machine])
        * (reward - Q[selected_machine])
    )

    total_reward += reward

    print(f"Updated Q({selected_machine}) = {Q[selected_machine]:.2f}")

print("\n\n========== FINAL RESULTS ==========")

print("\nMachine\tN\tR\tQ")
print("-----------------------------")

for machine in machines:
    print(
        f"{machine}\t"
        f"{N[machine]}\t"
        f"{R[machine]:.2f}\t"
        f"{Q[machine]:.2f}"
    )

print("\nTotal Reward =", total_reward)

print("\nAnalysis:")

for machine in machines:
    print(
        f"Machine {machine}: selected {N[machine]} times, "
        f"total reward = {R[machine]:.2f}, "
        f"estimated Q-value = {Q[machine]:.2f}"
    )
```
