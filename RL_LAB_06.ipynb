```python
import gymnasium as gym


def run_random_agent(env_name, num_episodes=5, max_steps=100):
  env = gym.make(env_name)
  episode_rewards = []

  print(f'=== Environment: {env_name} ===')
  print(f'Action Space: {env.action_space}')
  print(f'Observation Space: {env.observation_space}\n')

  for episode in range(1, num_episodes + 1):
    state, info = env.reset()
    total_reward = 0
    steps = 0

    for step in range(max_steps):
      action = env.action_space.sample()
      next_state, reward, terminated, truncated, info = env.step(action)
      total_reward += reward
      state = next_state
      steps += 1

      if terminated or truncated:
        break

    episode_rewards.append(total_reward)
    print(f'Episode {episode}: Steps = {steps}, Total Reward = {total_reward}')

  avg_reward = sum(episode_rewards) / len(episode_rewards)
  print(f'\nAverage Reward over {num_episodes} episodes: {avg_reward:.2f}\n')
  env.close()


run_random_agent('CliffWalking-v1', num_episodes=5, max_steps=200)
run_random_agent('Taxi-v4', num_episodes=5, max_steps=200)
```
