```python
n = int(input("Enter number of machines: "))

machines = {}
Q = {}
N = {}

for i in range(n):
    name = input(f"Enter machine {i + 1} name: ")
    rewards = list(map(float, input("Enter rewards separated by spaces: ").split()))

    machines[name] = rewards
    Q[name] = 0.0
    N[name] = 0

for machine in machines:
    for reward in machines[machine]:
        N[machine] += 1
        Q[machine] = Q[machine] + (reward - Q[machine]) / N[machine]

print("\n--- Estimated Q Values ---")

for machine in machines:
    print(f"\nMachine {machine}")
    print(f"Rewards: {machines[machine]}")
    print(f"Q({machine}) = {Q[machine]:.2f}")

selected_machine = max(Q, key=Q.get)

print(f"\nGreedy Selected Machine: {selected_machine}")
```
